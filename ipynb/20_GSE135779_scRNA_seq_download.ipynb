# 20 · GSE135779 · scRNA_seq · download

Nehar-Belaid D et al. Mapping systemic lupus erythematosus heterogeneity at the single-cell level.
*Nat Immunol* 2020;21:1094–1106. PBMC (peripheral blood mononuclear cells), 10x Genomics.
33 children with SLE and 11 healthy children (plus adults, not used).

Raw reads are controlled access (dbGaP phs002048). The per-donor count matrices are public on GEO.
GEO holds only age, age group and SLE/healthy; the clinical data are in the paper's
Supplementary Table 1.

| file | source | size |
|---|---|---|
| `GSE135779_RAW.tar` | GEO supplementary: one count matrix per donor | 1.3 GB |
| `GSE135779_genes.tsv.gz` | GEO supplementary: genes of the matrices | 0.3 MB |
| `GSE135779_series_matrix.txt.gz` | GEO series matrix | 5 KB |
| `41590_2020_743_MOESM3_ESM.xlsx` | *Nature Immunology* article page, Supplementary Tables 1–4, downloaded by hand | 0.46 MB |

```r
source("../src/paths.R")
options(timeout = 7200)
base <- "https://ftp.ncbi.nlm.nih.gov/geo/series/GSE135nnn/GSE135779"
files <- c(tar = raw("GSE135779", "GSE135779_RAW.tar"),
           genes = raw("GSE135779", "GSE135779_genes.tsv.gz"),
           series = raw("GSE135779", "GSE135779_series_matrix.txt.gz"),
           supplement = raw("GSE135779", "41590_2020_743_MOESM3_ESM.xlsx"))
# The publisher's and PubMed Central's sites answer scripted requests with a check page, so the
# supplement is downloaded by hand in a browser (see below) and is not fetched here.
urls <- c(tar = file.path(base, "suppl/GSE135779_RAW.tar"),
          genes = file.path(base, "suppl/GSE135779_genes.tsv.gz"),
          series = file.path(base, "matrix/GSE135779_series_matrix.txt.gz"))
for (f in c("tar", "genes", "series")) if (!file.exists(files[[f]])) download.file(urls[[f]], files[[f]], mode = "wb")
data.frame(file = basename(files), MB = round(file.size(files) / 1e6, 2))
```

**The supplement, by hand.** Open https://www.nature.com/articles/s41590-020-0743-0 in a browser,
download Supplementary Tables 1–4 (`41590_2020_743_MOESM3_ESM.xlsx`) and save it in `data/GSE135779/`.

**Check.** The file must exist and be an Excel file (a zip archive, first two bytes `PK`), not a web page.

```r
supplement_ok <- file.exists(files[["supplement"]]) &&
  identical(rawToChar(readBin(files[["supplement"]], "raw", 2)), "PK")
stopifnot(supplement_ok)
readxl::excel_sheets(files[["supplement"]])
```

```r
rawdir <- raw("GSE135779", "raw")
if (!length(list.files(rawdir, pattern = "matrix.mtx.gz$"))) { dir.create(rawdir, showWarnings = FALSE); untar(files[["tar"]], exdir = rawdir) }
c(matrices = length(list.files(rawdir, pattern = "matrix.mtx.gz$")),
  barcode_files = length(list.files(rawdir, pattern = "barcodes.tsv.gz$")))
```

**Result.** 56 per-donor matrices (one per GEO sample), and the supplement with 23 sheets.
