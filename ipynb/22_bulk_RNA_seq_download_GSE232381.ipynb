# Step 22 — download GSE232381, bulk RNA sequencing

**Data type: bulk_RNA_seq.** Illumina NovaSeq 6000, peripheral blood cells. **Reads:** nothing local.
**Writes:** `data/GSE232381/`.

**Source.** Chen YC et al. *Heliyon* 2024;10(11):e32303, doi:10.1016/j.heliyon.2024.e32303. Lupus
nephritis: active against inactive disease. GEO lists 25 samples (12 active, 13 inactive).

**The counts.** The authors did not deposit a counts table; their supplementary file is a spreadsheet
of results. GEO's own RNA sequencing pipeline, however, has re-aligned the raw reads and publishes
**raw integer counts per gene** (GRCh38.p13, keyed by NCBI Entrez gene identifier). That table covers
**16 of the 25 samples**: GEO had not processed the other nine. We use those 16.

GEO's companion gene annotation file sits behind an interactive check and cannot be fetched by a
script. We map identifiers with NCBI's `Homo_sapiens.gene_info.gz` instead (step 00), which is the
same source.

```r
source("../src/paths.R")
COUNTS <- fetch("https://www.ncbi.nlm.nih.gov/geo/download/?type=rnaseq_counts&acc=GSE232381&format=file&file=GSE232381_raw_counts_GRCh38.p13_NCBI.tsv.gz",
                raw("GSE232381", "GSE232381_raw_counts_GRCh38.p13_NCBI.tsv.gz"))
SERIES <- fetch("https://ftp.ncbi.nlm.nih.gov/geo/series/GSE232nnn/GSE232381/matrix/GSE232381_series_matrix.txt.gz",
                raw("GSE232381", "GSE232381_series_matrix.txt.gz"))
# guard against an HTML page saved in place of the table
stopifnot(identical(readBin(COUNTS, "raw", 2), as.raw(c(0x1f, 0x8b))))
data.frame(file = basename(c(COUNTS, SERIES)), KB = round(file.size(c(COUNTS, SERIES)) / 1e3))
```

## Findings

Both files are on disk: a gzip-compressed counts table and the series matrix with the sample labels.
