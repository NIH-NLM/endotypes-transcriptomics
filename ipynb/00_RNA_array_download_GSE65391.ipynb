# Step 00 — download GSE65391 and the gene annotation

**Data type: RNA_array.** Illumina HumanHT-12 v4 microarray (GEO platform GPL10558), whole blood.
**Reads:** nothing local. **Writes:** `data/GSE65391/` and `data/ncbi/`.

**Source.** Banchereau R, et al. Personalized immunomonitoring uncovers molecular networks that
stratify lupus patients. *Cell* 2016;165(3):551–565. doi:10.1016/j.cell.2016.03.008.
GEO (Gene Expression Omnibus) accession **GSE65391**: 158 children with systemic lupus erythematosus
(SLE), 924 whole-blood samples, and pediatric healthy controls.

Three files, each downloaded once. A second run touches no network.

| file | what it is |
|---|---|
| `GSE65391_series_matrix.txt.gz` | expression values and sample characteristics |
| `GPL10558.soft.gz` | the array annotation: which gene each probe measures |
| `Homo_sapiens.gene_info.gz` | NCBI's gene table, used to give every study the same gene names |

```r
source("../src/paths.R")
source("../src/platforms.R")
suppressMessages(library(GEOquery))

SERIES <- fetch("https://ftp.ncbi.nlm.nih.gov/geo/series/GSE65nnn/GSE65391/matrix/GSE65391_series_matrix.txt.gz",
                raw("GSE65391", "GSE65391_series_matrix.txt.gz"))
GPL <- raw("GSE65391", "GPL10558.soft.gz")
if (!file.exists(GPL)) invisible(getGEO("GPL10558", destdir = raw("GSE65391")))
gi  <- read_gene_info()                                  # downloads data/ncbi/ on first use

files <- c(SERIES, GPL, file.path(NCBI, "Homo_sapiens.gene_info.gz"))
data.frame(file = basename(files), MB = round(file.size(files) / 1e6, 1))
```

```r
c(genes_in_ncbi_gene_info = nrow(gi), with_ensembl_id = sum(!is.na(gi$ensembl)))
```
