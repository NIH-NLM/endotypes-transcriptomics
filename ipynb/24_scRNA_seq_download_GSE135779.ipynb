# Step 24 — download GSE135779, single-cell RNA sequencing

**Data type: scRNA_seq.** 10x Genomics, peripheral blood mononuclear cells (PBMC).
**Reads:** nothing local. **Writes:** `data/GSE135779/`.

**Source.** Nehar-Belaid D et al. Mapping systemic lupus erythematosus heterogeneity at the single-cell
level. *Nat Immunol* 2020;21:1094–1106, doi:10.1038/s41590-020-0743-0. 33 children with SLE and 11
healthy children (plus adults, whom we do not use). Raw sequencing reads are controlled access
(dbGaP phs002048), but **per-cell count matrices for every donor are public on GEO**. That is all we
need for pseudobulk.

**The download is 1.2 GB**, one archive of 56 per-donor matrices. It is fetched once and unpacked into
`data/GSE135779/raw/`.

```r
source("../src/paths.R")
BASE <- "https://ftp.ncbi.nlm.nih.gov/geo/series/GSE135nnn/GSE135779"
TAR    <- fetch(file.path(BASE, "suppl/GSE135779_RAW.tar"),  raw("GSE135779", "GSE135779_RAW.tar"))
GENES  <- fetch(file.path(BASE, "suppl/GSE135779_genes.tsv.gz"), raw("GSE135779", "GSE135779_genes.tsv.gz"))
SERIES <- fetch(file.path(BASE, "matrix/GSE135779_series_matrix.txt.gz"), raw("GSE135779", "GSE135779_series_matrix.txt.gz"))
RAWDIR <- raw("GSE135779", "raw")
if (!length(list.files(RAWDIR, pattern = "matrix.mtx.gz$"))) {
  dir.create(RAWDIR, showWarnings = FALSE)
  untar(TAR, exdir = RAWDIR)
}
c(archive_GB = round(file.size(TAR) / 1e9, 2),
  matrices = length(list.files(RAWDIR, pattern = "matrix.mtx.gz$")),
  barcode_files = length(list.files(RAWDIR, pattern = "barcodes.tsv.gz$")))
```

## Findings

56 per-donor matrices are unpacked, one per GEO sample, sharing one gene list.
